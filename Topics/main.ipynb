## Python Errors

Exceptions are unexpected events that could happen during the execution of code. In python `exceptions` are used symnomonously with `errors` that are raised/thrown by the code. A raised error can be caught by the handler depending on the context, if not then the error would interrupt and stop the code execution.

#### Basic / Common Errors:

<th>Error Type</th>
<th>Meaning</th>

<tr><td>ValueError</td> <td>Right type of data, but weird value</td> <td>Setting your age to -5</td></tr>
