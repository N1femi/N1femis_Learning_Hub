## Python Errors

Exceptions are unexpected events that could happen during the execution of code. In python `exceptions` are used symnomonously with `errors` that are raised/thrown by the code. A raised error can be caught by the handler depending on the context, if not then the error would interrupt and stop the code execution.

#### Basic / Common Errors:
<table>
<th>Class</th>
<th>Description</th>

<tr><td>Exception</td> <td>Base class for most error types</td></tr>
<tr><td>AttributeError</td> <td>Raised by syntax obj.method if obj has no attribute named "method"</td></tr>
<tr><td>TypeError</td> <td>Raised when wrong parameter is sent to a function</td></tr>
<tr><td>ValueError</td> <td>Raised when parameter is an invalid value</td></tr>
<tr><td>IndexError</td> <td>Raised if index to sequence is out of bounds</td></tr>
<tr><td>KeyError</td> <td>Raised is non-existent key was requested from set or dictionary</td></tr>
<tr><td>NameError</td> <td>Raised if nonexistent identifier was used</td></tr>
<tr><td>ZeroDivisionError</td> <td>Raised when attempting to divide by 0 (don't break math)</td></tr>
<tr><td>EOFError</td> <td>Raised if end of file is reached for console or file input</td></tr>
<tr><td>IOError</td> <td>Raised upon different types of failure when dealing with file I/O operations</td></tr>
<tr><td>StopIteration</td> <td>Raised by next(iterator) is no element is next</td></tr>

</table>
