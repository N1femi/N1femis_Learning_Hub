# Dictionaries

Mutuable structures structured as a `key:value` relationship.

Take the following dictionary as an example:

```python
fruits = {
    "Apple": 2,
    "Strawberry": 8,
    "Banana": 5,
    "Rasberry": 14,
    "Blueberry": 29
}
```

In the above dictionary named **fruits**, see how each line follows a `key: value` syntax, with commas *seperating* each pair. In this example the keys are the fruits with the values corresponding to how many of each we have.

Unlike tuples/lists we are **NOT** able to index through dictionaries, meaning running `fruits[0]` will not work. Instead, we must explicitely use keys. Here is an example:

```python
fruits["Apple"] # 'Apple' is a key and should return 2 as its corresponding value
```

## Methods
Below I will list the useful methods that complement data structure knowledge and are overall very beneficial to know. When explaining I will refer to the dictionary as `dict.method()` so you understand the general sense in which these methods are applied, however the examples will be using the `fruits` dictionary.

`dict.get(key)` is a safer way to retrieve a key, in most cases it's an alternative for `dict[key]` **BUT** if there's a case where the *key* doesn't exist in the dictionary: `dict[key]` will raise a <a href="../Errors/main.ipynb">Key Error</a>, on the other hand `dict.get(key)` will not.

```python
fruits.get("Apple")
```

```python
all_fruits = list(fruits.keys())
print(all_fruits)
```

```python
all_amounts = list(fruits.values())
print(all_amounts)
```

```python
all_fruit_pairs = list(fruits.items())
print(all_fruit_pairs)
```
