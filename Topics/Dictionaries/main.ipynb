# Dictionaries

Mutuable structures structured as a `key:value` relationship.

Take the following dictionary as an example:

```python
fruits = {
    "Apple": 2,
    "Strawberry": 8,
    "Banana": 5,
    "Rasberry": 14,
    "Blueberry": 29
}
```

In the above dictionary named **fruits**, see how each line follows a `key: value` syntax, with commas *seperating* each pair. In this example the keys are the fruits with the values corresponding to how many of each we have.

Unlike tuples/lists we are **NOT** able to index through dictionaries, meaning running `fruits[0]` will not work. Instead, we must explicitely use keys. Here is an example:

```python
fruits["Apple"] # 'Apple' is a key and should return 2 as it's corresponding value
```
