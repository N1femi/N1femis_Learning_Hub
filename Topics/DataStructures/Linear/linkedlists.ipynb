# Linked Lists

Contains ordered Nodes where the first Node(called the **head**) starts the list, and the last Node (called the **tail**) points to `None`

Linked Lists are a dynamic data structures which means they can **grow & shrink** during runtime. So there is no need for an initial size.

However, Linked Lists have a linear Lookup time since you would have to find any value by starting at the head and chaining `.next`'s


Here's an example of a LinkedList, starting with the foundation—the NODE!

```python
class Node():
    def __init__(self, value):
        self.value = value
        self.next = None
```

### Now the Singly LinkedList Class

Each LinkedList object will only contain reference to the **head** object, because we can find any other value in the list by using `head.next`. 

Notes: 

LinkedList.delete() <u>DOES NOT directly</u> delete the initial head, it removes the referenece which then allows **garbage collector** to finish the job >:)

```python
class LinkedList():
    def __init__(self):
        self.head = None
    
    def add(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node
    
    def delete(self):
        if self.head is None:
            return ("The list is already empty brochacho.")
        self.head = self.head.next
```

## Doubly-Linked List

Just a regular LinkedList with an additional pointer called **prev**.

Due to this additional pointer we are able to traverse forwards <u>AND</u> backwards. However, every Node now require extra space for a new pointer, and we will have to keep that in mind for its operations.

## Circular Linked List

This is a LinkedList where all nodes are connected to form a circle. This means any node could be a starting point and there is no way for `Node.next == None`.

When adding a new node we will have to set its **head** pointer to the new node and `last.next` to the new node.

These are particularly useful for <a href="./Stacks&Queues.ipynb">queues</a>.
