# Stack

A stack is an ordered collection of items where insertion and deletion of items occur on the same end.

Elements are inserted an removed adhering to the LIFO (Last in First out)

The **main methods** of stacks are:
- **Push**: Adds element to Stack

- **Pop**: Removes element from Stack
- **isEmpty**: Returns boolean based on if stack is Empty
- **Peek**: Returns top element of Stack
- **Size**: Returns number of items in Stack

Here is a Stack's Implementation using **Nodes**:

First we start with the building block (Node Class)

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None 
```

Now the actual Stack Class

```python
class Stack:
    def __init__(self):
        self.top = None
    
    def push(self, value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node
        
    def pop(self):
        popped_node = self.top
        self.top = self.top.next
        
        return popped_node.value
    
    def isEmpty(self):
        return self.top is None
```

### *Application of Stacks*

Stacks can be used to evaluate **Infix** & **Postfix** expressions.

- **Infix**: Math written with the operator between the two operands. *Ex: (a + b)*

- **Postfix**: Math written with the operator following the operands. *Ex: (a b +)*

Let's try it below:
A stack will help us solve these expressions by,

1. `Pushing` encountered operands into the Stack

2. If we encounter an operation, we `pop()` and use the last two operands in the Stack

3. After doing some math, we push the result back into the Stack

4. We <u>repeat</u> this process until the Stack eventually has one item left.

Note: This is possible because of how Postfix has order encoded, hence why parenthesis aren't required. <a href="https://www.geeksforgeeks.org/dsa/evaluation-of-postfix-expression/">Learn more here</a>

```python
import operator
# This built-in library will allow us to make a dictionary of operators matching with their functioning

#Infix Version: 4 + (3 - 2) * 5
expression = "432-5*+"

operators = {
    '*': operator.mul, 
    '-': operator.sub, 
    "+": operator.add, 
    "/": operator.truediv
    }

myStack = Stack()

for item in expression:
    if item in operators:
        num2 = myStack.pop()
        num1 = myStack.pop()
        
        operator_func = operators[item] # This is why we used operators library
        result = operator_func(num1, num2)
        myStack.push(result)
    else:
        myStack.push(int(item))

print(myStack.top.value)
# We should get 9

```
