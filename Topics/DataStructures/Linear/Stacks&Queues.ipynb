# Stack

A stack is an ordered collection of items where insertion and deletion of any item happens on the same end.

Elements are inserted an removed adhering to the LIFO (Last in First out)

The **main methods** of stacks are:
- Push
- Pop
- isEmpty
- Peek
- Size

Here is how it can be created from a python list:

```python
class Stack():
    def __init__(self):
        self.items = []
    
    def push(self, item):
        self.items.append(item)
    
    def pop(self):
        return self.items.pop()
    
    def Peek(self):
        return self.items[len(self.items) - 1]
    
    def isEmpty(self):
        return self.items == []
    
    def size(self):
        return len(self.items)
    
```
