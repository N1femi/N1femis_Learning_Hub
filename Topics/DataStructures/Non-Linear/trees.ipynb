# Trees

A tree is a heirarchical data structure with a Parent-to-child relationship. 

A tree is:
- composed of **nodes**(which hold the data) & **edges**(which connects them with other nodes)
- does not have a cycle

The bottom nodes of a tree are called **leaf/external nodes**

##### 3 <u>Important Properties:</u>
- **Height**: Number of edges from Node to lowest leaf
    - *Height* of Tree = *Height* of Root Node

- **Degree**: Number of childen connected to evaluated Node
- **Depth**: Number of edges from root to evaluated Node
- **Level**: The depth of node from the root (*Level = Depth + 1*)
- **Edge**: A connection between one node to another
- **Path**: A sequence of nodes & edges connecting to a descendant

### Binary Tree
In this type of tree every node has at most **2 children**. *Typically named "left" and "right"*

There are **4 types** of binary trees:
- **Full Type:** Every node has <u>0</u> or <u>2</u> children

- **Complete Type:** Every level is fully filled except possibly the last, which fills left to right with no gaps.
- **Perfect Type:** Every node had 2 children, all leaves are on same level
- **Degenerate Type:** Every node only has 1 child, which could behvave as a <a href="../Linear/linkedlists.ipynb">Linked List</a>

*Note: A Perfect Tree is also: Complete & Full*


Properties of binary trees:
<br>
$n$ = Nodes
<br>
$k$ = Levels

1. Max children per Node: 2
2. Nodes per level: $\leq 2^{k-1}$
3. Total Nodes: $\leq 2^k - 1$
4. Total Edges: $n - 1$


Now let's implement a Binary Tree starting with it's Node Foundation:




```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
```

We will be creating a special type of Binary Tree: **Binary Search Tree**

In this tree, for every node:
- All values in the left subtree are **smaller**
- All values in the right subtree are **bigger**

*For **EVERY** Node: Left < Node < Right*

```python
class BinarySearchTree:
    def __init__(self):
        self.root = None
        
    def insert(self, node, value): # Could loop too, but convention is recursion
        if node == None:
            self.root = Node(value) # If no Initial Node, the value becomes the root Node
            
        else:
            if value < node.value:
                if node.value.left == None: # Smaller values go on left
                    node.value.left = Node(value)
                else:
                    self.insert(node.left, value)
            if value > node.value:
                if node.value.right == None: # Larger values go on right
                    node.value.right = Node(value)
                else:
                    self.insert(node.right, value)
            
```

## Binary Tree Traversal

Types of Traversals:
- **Breadth-First(BFS) / Level-Order Search:** Visits every node on a level, from left to right, Dbefore going to lower level.

- **Depth-First(DFS) Search:** Goes deep as possible down one path before backing up and trying a different one.
    -  **Preorder Traversal:** Root, Left, Right
    -  **Inorder Traversal:** Left, Root, Right
    -  **Postorder Traversal:** Left, Right, Root


### Binary Heap
A binary heap has these properties:
- **Complete Binary Tree** Structure
- Heap Ordering Propert:
    - Min-Heap(every child is greater than or equal to its parent)
    - Max-Heap(every child is less than or equal to its parent)

Each path is sorted but subtrees are not sorted relative to each other
Binary Heap **DOES NOT** equal a Binary Search Tree
