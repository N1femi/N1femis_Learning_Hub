# Trees

A tree is a heirarchical data structure with a Parent-to-child relationship. 

A tree is:
- composed of **nodes**(which hold the data) & **edges**(which connects them with other nodes)
- does not have a cycle

The bottom nodes of a tree are called **leaf/external nodes**

##### 3 <u>Important Properties:</u>
- Height: Number of edges from root -> lowest leaf
    - *Height* of Tree = *Height* of Root Node

- Degree: Number of childen connected to evaluated Node
- Depth: Number of edges from root to evaluated Node
- Level: **Depth** + 1
- Edge: A connection between one node to another
- Path: A sequence of nodes & edges connecting to a descendant

### Binary Tree
In this type of tree every node has at most **2 children**. *Typically named "left" and "right"*

There are **4 types** of binary trees:
- **Full Type:** Every node has <u>0</u> or <u>2</u> children

- **Complete Type:** Every level is fully filled except possibly the last, which fills left to right with no gaps.
- **Perfect Type:** Every node had 2 children, all leaves are on same level
- **Degenerate Type:** Every node only has 1 child, which could behvave as a <a href="../Linear/linkedlists.ipynb">Linked List</a>

*Note: A Perfect Tree is also: Complete & Full*


##### Properties of binary trees
- A tree of **n** nodes has **(n - 1) edges**

- For every **(k >= 1)** there are atmost **2^(k-1) nodes** in **level k**
