# Recursion 
---

Recursion is when a function calls itself, as shown:

```python
def giveNumber(num):
    return giveNumber(num)   
```

The function `giveNumber(num)` calls itself during its execution, but **why even do this?**

Recursion is built on the principle: Breaking a problem into smaller, more simple parts to reach a simple **base case**

Every Recursive function needs **TWO** things or it'll run forever:
- **Base Case:** A condition which it'll <u>stop</u> calling itself
- **Recursive Case:** A condition which it'll call itself with a smaller input

A good concept to model a recursive function after are factorials:

Factorials (written as x!): Means multiply all positive integars from 1 to x.
5! = 5 x 4 x 3 x 2 x 1 = 120

(Note: Both 0! and 1! equates to 1)

```python
def factorial(x):
    # BASE CASE: smallest problem we can solve directly
    if x == 0 or x == 1: 
        return 1
    else:
        # RECURSIVE CASE: break into smaller problems
        return x * factorial(x - 1)
```
