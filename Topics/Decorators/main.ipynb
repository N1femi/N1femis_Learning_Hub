# Decorators
---
Extends the behavior of functions or classes **without** modifying the base code.



Now, we will make our own decorator:

```python
def add_sprinkles(func): # This is the decorator name
    def wrapper(): # The wrapper mentioned above
        print("Adding Sprinkles...")
        
        func()
    
    return wrapper # Return the Wrapper

# @add_sprinkles <-- This is our custom decorator
```

`@add_sprinkles` is syntactic sugar which means it is a minimal formatting method synonymous to passing the function in manually: `add_sprinkles(get_ice_cream)`

```python
@add_sprinkles
def get_ice_cream():
    print("Here is your ice cream ")
    
get_ice_cream()
```

The `wrapper()` function is needed because without it `@add_sprinkles` would be calling the function immediately.
