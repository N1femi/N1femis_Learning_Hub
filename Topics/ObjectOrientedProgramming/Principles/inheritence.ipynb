## Inheritence 🧔‍♂️ ➡️ 🧒
---

To start this demonstration we will create a **class** named `Animal`:

```python
class Animal():
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age
    
    def make_sound(self):
        return "Some generic animal sound"
```

If you are still not familiar with how to create classes or the syntax please <a href="../main.ipynb">click here</a>

Now that we have the *Animal* class, we will create **2 Child Classes**:

```python
class Dog(Animal): # Child Class 1
    def __init__(self, breed, name, age):
        super().__init__(name, "Canis lupus familiaris", age)
        self.breed = breed
        

class Cat(Animal): # Child Class 2
    def __init__(self, indoors, name, age):
        super().__init__(name, "Felis catus", age)
        self.indoors = indoors
```

Now that we created **2 child classes (Dog, and Cat)**, what inheritance allows us to do is basically have all of the attributes & methods the <u>parent</u> class had. 

To reference the parent class, we called `super()` then we call the `__init__()` method. Doing this, we are able to "inherit" the parent's attributes. On the other hand, methods are automatically inherited.

Let's see what we can do with our new child classes:

```python
myCat = Cat(True, "Kitty", 2)
print(f"My cat's name is {myCat.name}, {myCat.age} years old, and is currently {'indoors' if myCat.indoors else 'outside'}.")

myDog = Dog("Golden Retriever", "Rufus", 3)
print(f"I have a dog named {myDog.name}, {myDog.age} years old, and is a {myDog.breed} breed.")
```
