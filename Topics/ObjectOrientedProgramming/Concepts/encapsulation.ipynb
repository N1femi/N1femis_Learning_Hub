# Encapsulation

Encapsulation allows us to bundle data (methods & attributes) to control how it's used. This is useful in situtations where we dont want users to **change data directly**, maybe because we want to maintain integrity and synchronize different parts of the object.

*I'll model a clearer example below:*

```python
# This class will be used as main example

class BankAccount():
    def __init__(self, acc_num, name):
        self.__account_number = acc_num
        self.name = name
        
    def reassign_account(self, new_num):
        self.__account_number = new_num
```

Now watch what happens when I print `name` and `__account_number` from a new object:

```python
new_account = BankAccount(25671, "Nifemi")

print(new_account.name)

print(new_account.__account_number) 
```
