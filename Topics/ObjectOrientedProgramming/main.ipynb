# Object-Oriented Programming (OOP)
---
Classes are the foundations of OOP. A **Class** is like a blueprint for building something, it defines what **attributes** (variables) and **methods** (functions) the object will have. The **Object** is the finished building created from that blueprint.

Here is a quick example of what a class looks like:

```python
class Person():
    def __init__(self, name, age, major):
        self.name = name # Attribute
        self.age = age # Attribute
        self.major = major # Attribute
    
    def introduce(self): # Method
        return f"Hi, I'm {self.name}, majoring in {self.major}."
    

Person1 = Person("Jake", 18, "Communications")
print(Person1.introduce())

Person2 = Person("Matthew", 19, "Mathematics")
print(Person2.introduce())
```

##### Let's break down the above class and label the different parts:
- Class: `Person()` 
- Attributes: `name`, `age`, `major`
- Methods: `introduce()`

- Objects: `Person1`, `Person2`

But wait, **what is `self`?**

Think of `self` as a way for an object to differentiate itself from others. For example: when a waiter in a resturant says "your order", the object knows if waiter is talking to them or different object.

##### There are <u>4 Principles of OOP</u>:
1. <a href="./Principles/inheritence.ipynb">**Inheritence**</a>: A child class *inherits* the attributes and methods of a specified Parent class.

2. <a href="./Principles/polymorphism.ipynb">**Polymorphism**</a>: The ability for a single object to take many forms

3. <a href="./Principles/encapsulation.ipynb">**Encapsulation**</a>: The bundling data and functions

4. <a href="./Principles/abstraction.ipynb">**Abstraction**</a>: Hiding complex implementation details and only showing the essential parts
