# Object-Oriented Programming (OOP)
---

#### <u>Classes</u> and <u>objects</u> are the two main aspects of OOP. A *class* creates a new type of instance, where *objects* are the instances created from that class.

Some notes:
- Variables that belong to an object/class are called *fields*
- Objects can use functions that belong to their class *(called methods)*
- Classes can be considered **organized collections of variables and functions**

Here is how you can define a class:

```python
class Person:
    random_var = 5
    
    def laugh(self):
        return ("Hahahahahahahaha")
    

N1femi = Person() # New object created from class "Person"

print(N1femi.random_var)
print(N1femi.laugh())
```

The `self` variable represents the object itself, using it within the class allows actions to affect the object instead of the entire class.

Also remember that `self` is skipped when calling a class because it is automatically passed as the first argument. Here is an example:

```python
class Person:
    def __init__(self, name):
        self.name = name
    
    def laugh(self):
        print(self.name + " is laughing...")

N1femi = Person('N1femi') # 'self' is skipped here since automatically passed in
N1femi.laugh()

Johnny = Person("Johnny")
Johnny.laugh()


    
```

In the above example I created two objects called "N1femi" and "Johnny" from the class *Person*. 

The class also has a method (`laugh()`) that can be called on each object. There is another method (`__init__()`), which will always be called when a new object is created. 

Once an object is created each of the same class have the same methods but variable contents could differ.

##### Context Terminology:
- `Person()`: A class which objects can be created from

- `N1femi`: An object created from the "Person" class

- `Johnny`: An object created from the "Person" class

- `.laugh()`: A class method shared by created objects

- `.name`: A variable different to each created object (*aka attribute*)
